```python
st={1,2,3,4,3,2,'hii',4.5,(3,4,2,4)}#we can't store duplicate value but if we are 
                                     #storing an tuple then it wont compare the main value with the value inside the tuple
st.remove(3)
print(st)
#if element is not preset then it will throw an error(solution->discard)
st.discard(4);
print(st);
```
